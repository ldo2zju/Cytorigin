Step 3. Load and preprocess datasets

CytOrigin currently only supports AnnData in a Python environment. Please convert a raw count matrix into AnnData objects if it is in other formats (e.g., Seurat, loom, SingleCellExperiment object)

```python
adata_A = sc.read_h5ad('/path/to/earlier_dataset')
adata_B = sc.read_h5ad('/path/to/later_dataset')
```

```python
# The reduction of the number of genes and cells is NOT strictly required. The number of genes can be increased as appropriate during actual generation.
adata = sc.concat([adata_A,adata_B],
                 join = 'outer',
                 fill_value=0,
                 index_unique='-')
sc.pp.filter_genes(adata, min_cells = 3)
sc.pp.highly_variable_genes(adata, n_top_genes = 8000)
adata = adata[:, adata.var.highly_variable]
adata
```

```python
adata.X = adata.layers['counts'].copy()
sc.pp.normalize_total(adata, target_sum=1e4)
```

```python
# to label the start point and end point of the two time points
adata.obs['s_e'] = adata.obs.time.map({'CS7':'start','CS9':'end'})
```
